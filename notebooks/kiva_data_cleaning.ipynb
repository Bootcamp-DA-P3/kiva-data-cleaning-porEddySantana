## Resumen de la limpieza

Durante el proceso de limpieza y validación se realizaron las siguientes comprobaciones y transformaciones:

- Se revisó la estructura inicial del dataset y sus tipos de datos.
- Se comprobaron valores nulos y cadenas vacías.
- Se verificaron filas duplicadas y duplicados en la columna `id`.
- Las columnas `posted_time`, `disbursed_time`, `funded_time` y `date` se convirtieron a formato `datetime`.
- `partner_id` y `term_in_months` se convirtieron a tipo entero.
- Se eliminaron espacios al inicio y al final de los campos `use` y `region`.
- Se comprobó la consistencia entre `country` y `country_code`.
- Se revisaron los valores de `funded_amount`, `loan_amount`, `term_in_months` y `lender_count`.
- Se verificaron valores extremos sin eliminar registros cuando no existía evidencia suficiente de que fueran errores.
- Se comprobó que `funded_amount` y `loan_amount` coinciden en todos los registros.
- Las columnas `tags` y `borrower_genders` se conservaron porque contienen información múltiple dentro de una misma celda.
- No se eliminaron filas del dataset.
